```python
import whitebox
import pathlib
wbt = whitebox.WhiteboxTools()
wbt.set_working_dir(pathlib.Path().resolve())

path_dem = "dem.tif"

def render_dme(scale_str, scale):
    dme_path = f'dme_{scale_str}.tif'
    dme_abs_path = f'dme_abs_{scale_str}.tif'
    wbt.dev_from_mean_elev(path_dem,
                           dme_path,
                               filterx=scale,
                               filtery=scale)

    wbt.absolute_value(
        dme_path,
        dme_abs_path
    )

render_dme('micro', 100)
render_dme('meso', 1000)
render_dme('macro', 10000)
```
